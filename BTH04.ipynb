```python
#Bài 4
import pandas as pd

#Tạo series
s1 = pd.Series([20, 17, 32, 19, 65])
print(s1)

print("==========")

#Thêm tên cho các dòng series
s1.index.name = "Thứ tự"
print(s1)
print("==========")

#Lấy giá trị ở dòng 2
print("Lấy giá trị ở dòng 2: " + str(s1[2]))
print("==========")
#Lấy giá trị 3 dòng cuối
print("Lấy giá trị 3 dòng cuối: ")
print(s1.tail(3))
```

```python
#Bài 5:
#Tạo ra một list để lưu trữ các dữ liệu trên
data_list = [
    [111, "Hong", 20],
    [222, "Tai", 19],
    [333, "Tam", 23],
    [444, "Phuong", 22],
    [555, "Huy", 21]
]

#Tạo ra một dataframe từ list ở trên
df = pd.DataFrame(data_list)

#Thêm tên dòng cho dataframe vừa tạo: SV1, SV2, SV3, SV4, SV5 (Sử dụng tham số index)
#Thêm tiêu đề cho dataframe: MSSV, HoTen, Tuoi (Sử dụng tham số columns)
df.index = ["SV1", "SV2", "SV3", "SV4", "SV5"]
df.columns = ["MSSV", "HoTen", "Tuoi"]

print("--- Dataframe sau khi đặt index và columns: ---")
print(df)
print("\n" + "="*40 + "\n")

#Hiển thị toàn bộ thông tin của sinh viên 2
thong_tin_sv2 = df.loc["SV2"]
print("--- Toàn bộ thông tin của sinh viên 2: ---")
print(thong_tin_sv2)
print("\n" + "="*40 + "\n")

#Hiển thị tên của tất cả sinh viên
ten_sinh_vien = df["HoTen"]
print("--- Tên của tất cả sinh viên: ---")
print(ten_sinh_vien)
print("\n" + "="*40 + "\n")

#Tạo một dataframe mới với thông tin bao gồm: MSSV và HoTen
df_moi = df[["MSSV", "HoTen"]]
print("--- Dataframe mới (chỉ gồm MSSV và HoTen): ---")
print(df_moi)
```

```python
#Bài 6:

#Sử dụng một dictionary để lưu danh sách sinh viên như trên
data_dict = {
    "MSSV": [101, 102, 103],
    "HoTen": ["Ly Thi Hong", "Nguyen Van Hai", "Le Phuong Anh"],
    "NamSinh": [2001, 2002, 2001],
    "DiaChi": ["HCM", "VT", "DN"]
}

#Tạo dataframe từ dictionary
df = pd.DataFrame(data_dict)

print("--- DataFrame ban đầu: ---")
print(df)
print("\n" + "="*50 + "\n")

#Thêm một cột dữ liệu "GioiTinh" vào danh sách trên
df["GioiTinh"] = ["Nu", "Nam", "Nu"]

#Thêm dữ liệu bên dưới vào danh sách sinh viên
data_them = {
    "MSSV": [104, 105, 106],
    "HoTen": ["Nguyen Van Tai", "Le Van Nam", "Phan Hong Tham"],
    "NamSinh": [2000, 2002, 2001],
    "DiaChi": ["NT", "VT", "HCM"],
    "GioiTinh": ["Nam", "Nam", "Nu"]
}
df_them = pd.DataFrame(data_them)

#Sử dụng pd.concat để gộp dòng
df = pd.concat([df, df_them], ignore_index=True)

print("--- DataFrame sau khi thêm cột và nối dữ liệu mới: ---")
print(df)
print("\n" + "="*50 + "\n")

#Tạo một dataframe là danh sách các bạn có giới tính Nam
df_nam = df[df["GioiTinh"] == "Nam"]
print("--- Danh sách sinh viên NAM: ---")
print(df_nam)
print("\n" + "="*50 + "\n")

#Lọc ra danh sách các bạn sinh viên Nữ có địa chỉ ở HCM và đếm số lượng
df_nu_hcm = df[(df["GioiTinh"] == "Nu") & (df["DiaChi"] == "HCM")]
so_luong_nu_hcm = len(df_nu_hcm)

print("--- Danh sách sinh viên Nữ ở HCM: ---")
print(df_nu_hcm)
print(f"\n=> Có {so_luong_nu_hcm} bạn sinh viên trong danh sách này.")
print("\n" + "="*50 + "\n")

#Lưu danh sách các bạn nữ có địa chỉ ở HCM vào file .CSV
df_nu_hcm.to_csv("sinh_vien_nu_hcm.csv", index=False)
print("-> Đã lưu danh sách sinh viên nữ ở HCM vào file 'sinh_vien_nu_hcm.csv' thành công!")
```
