```python
#Bai 8: Tao mang
list = np.array(
    [
        [5, 10, 21],
        [11, 23, 18],
        [22, 19, 35],
        [96, 22, 26]
    ]
)
print(list)
#So chieu, so dong va so cot
print("So chieu: " + str(list.ndim))
print("So dong: " + str(list.shape[0]))
print("So cot: " + str(list.shape[1]))

#Lay gia tri dong 2
print("Gia tri dong 2: " + str(list[1]))
#Lay gia tri dong 3
print("Gia tri dong 3: " + str(list[2]))

#Lay gia tri cot 3
print("Gia tri cot 3: " + str(list[:, 2]))

#Tao ma tran moi voi cac gia tri 23, 18, 19, 35, 22, 26 lay tu ma tran tren
new_matrix = list[1:4, 1:3]
print("Ma tran moi: ")
print(new_matrix)

#Thay doi gia tri 22 o dong 4, cot 2 thanh gia tri 30
list[3, 1] = 30
print("Ma tran sau khi thay doi so 22 thanh 30: ")
print(list)
```

```python
#Bai 10: Mang 1 chieu voi cac gia tri 99, 46, 87, 62, 11, 32
vector_9 = np.array([99, 46, 87, 62, 11, 32])
print(vector_9)
#Tim gia tri lon nhat va nho nhat tren mang
print("Gía trị lớn nhất: " + str(np.max(vector_9)))
print("Gía trị nhỏ nhất: " + str(np.min(vector_9)))
#Sắp xếp mảng theo thứ tự tăng dần
print("Mảng sau khi sắp xếp tăng dần: ")
print(np.sort(vector_9))
#In ra các giá trị ở vị trí chẵn của mảng
print("Giá trị ở vị trí chẵn: ")
i = 0
while i < len(vector_9):
  if(i == 0 or i % 2 == 0):
    print(vector_9[i])
  i = i + 1

#Kiem tra phan tu thu 4 la chan hay le. Xuat thong bao ket qua
if (vector_9[3] % 2 == 0):
  print("Phần tử thứ 4 là số chẵn")
else:
  print("Phần tử thứ 4 là số lẻ")

#Đếm số phần tử chẵn trong mảng
count = 0
j = 0
while j < len(vector_9):
  if (vector_9[j] % 2 == 0):
    count = count + 1
  j = j + 1
print("Số phần tử chẵn trong mảng là: " + str(count))

#Tính giá trị trung bình, giá trị trung vị và tổng các phần tử của mảng
print("Giá trị trung bình: " + str(np.mean(vector_9)))
print("Giá trị trung vị: " + str(np.median(vector_9)))
print("Tổng các phần tử: " + str(np.sum(vector_9)))

#Thay đổi giá trị 87 thành giá trị trung bình của mảng
vector_9[2] = np.mean(vector_9)
print("Mảng sau khi thay đổi giá trị 87 thành giá trị trung bình: ")
print(vector_9)

#Xóa phần tử có giá trị 62 trong mảng
new_vector_9 = np.delete(vector_9, np.where(vector_9 == 62))
print("Mảng sau khi xóa phần tử có giá trị 62: ")
print(new_vector_9)
```

```python
#Bài 11: Ma trận 3 chiều
matrix_3d = np.array(
    [
        [
           [5, 10, 21],
           [11, 23, 18],
           [22, 19, 35]
        ],
        [
            [50, 10, 41],
           [61, 23, 28],
           [42, 11, 25]
        ],
        [
            [96, 22, 26],
           [87, 56, 23],
           [90, 75, 32]
        ]
    ]
)
print(matrix_3d)

#Truy xuất hai dòng cuối của ma trận thứ nhất
print("Hai dòng cuối của ma trận thứ nhất: ")
print(matrix_3d[0, 1:3, :])

#Thay đổi các giá trị cột đầu tiên của ma trận thứ hai bằng giá trị trung bình của ma trận thứ nhất
mean_first_matrix = np.mean(matrix_3d[0])
matrix_3d[1, :, 0] = mean_first_matrix
print("Ma trận thứ hai sau khi thay đổi cột đầu tiên: ")
print(matrix_3d[1])

#Tính tổng giá trị của ma trận thứ ba
total_third_matrix = np.sum(matrix_3d[2])
print("Tổng các phần tử của ma trận thứ ba: " + str(total_third_matrix))

#Tạo ma trận mới từ ma trận thứ 2 và cho biết số chiều
new_matrix = matrix_3d[1]
print(new_matrix)
print("Số chiều của ma trận: " + str(new_matrix.ndim))

#Chuyển ma trận vừa tạo thành vector. Thêm vào cuối mảng giá trị 45
new_vector = new_matrix.flatten()
new_vector = np.append(new_vector, 45)
print("Vector sau khi thêm giá trị 45: ")
print(new_vector)

#Thêm vào đầu mảng giá trị 99
new_vector = np.insert(new_vector, 0, 99)
print("Vector sau khi thêm giá trị 99: " + str(new_vector))

#Tạo một list các giá trị được lấy từ ma trận 3 chiều ban đầu
list_matrix = matrix_3d.flatten()
print("List các giá trị được lấy từ ma trận 3 chiều ban đầu: ")
print(list_matrix)
```

```python
#Bài 13: sinhvien.txt

#Tạo file sinhvien.txt
noi_dung = """1 Thanh 20
2 Tam 21
3 Binh 22
4 Chinh 23
5 Lan 21"""

with open("sinhvien.txt", "w") as f:
    f.write(noi_dung)

#Đọc file vào ma trận (NumPy array) kiểu chuỗi
matrix = np.loadtxt("sinhvien.txt", dtype=str)
print("Ma trận dữ liệu:")
print(matrix)

#Thêm dòng tiêu đề và lưu thành file mới
tieu_de = "Stt HoTen Tuoi"
np.savetxt("sinhvien_tieude.txt", matrix, fmt="%s", header=tieu_de, comments="")

#Lấy cột Tuổi (cột chỉ số 2), ép kiểu sang số nguyên và tính trung bình
tuoi = matrix[:, 2].astype(int)
tuoi_tb = np.mean(tuoi)

print("Độ tuổi trung bình:", tuoi_tb)
```
